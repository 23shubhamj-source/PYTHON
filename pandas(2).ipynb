```python
import pandas as pd
info={"one":pd.Series([1,2,3,4,5],index=["a","b","c","d","e"]),"two":pd.Series([1,2,3,4,5,6],index=["a","b","c","d","e","f"])}
f=pd.DataFrame(info)
print("Add new columns by passing series")
f["three"]=pd.Series([20,40,60],index=["a","b","c"])
print(f)
print("Add new columns using existing DataFrame columns")
f["four"]=f["one"]+f["three"]
print(f)
```

```python
import pandas as pd
info={"one":pd.Series([1,2],index=["a","b"]),"two":pd.Series([1,2,3],index=["a","b","c"])}
f=pd.DataFrame(info)
print("The DataFrame:")
print(f)

#By del function:

print("Delete the first column:")
del f["two"]
print(f)

#By pop function:
print("Delete another column:")
f.pop("one")
print(f)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
#Load the csv dataset
a= pd.read_csv(r"C:\Users\Admin\Desktop\Stud.csv")
#Display the dataset
print("Dataset:")
print(a)
print(a.isnull())

#Line plot
plt.figure(figsize=(7,6))
plt.plot(a["students"],a["python"],marker="o")
plt.title("students vs python")
plt.xlabel("students")
plt.ylabel("python")
plt.show()

#Line plot
plt.figure(figsize=(5,5))
plt.plot(a["students"],a["maths"], marker="d")
plt.title("students vs maths")
plt.xlabel("students")
plt.ylabel("maths")
plt.show()

#Bar plot
plt.figure(figsize=(7,6))
plt.bar(a["students"],a["python"])
plt.title("Python marks of students")
plt.xlabel("students")
plt.ylabel("python")
plt.show()
```

```python
#Scatter plot
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

a=pd.read_csv(r"C:\Users\Admin\Desktop\Stud.csv")
plt.figure(figsize=(6,6))
plt.title("Student and Python")
plt.scatter(a["students"],a["python"],  s= 50)
plt.xlabel("students")
plt.ylabel("python")
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
